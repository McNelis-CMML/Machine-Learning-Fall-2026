## Core Exercise 3 — GJR-GARCH-in-Mean

Using `ARCHInMean`, estimate an AR(5) GARCH-in-Mean model with

`GARCH(p=1, o=1, q=1)`

and Student-\(t\) innovations.

1. Print the summary.
2. Identify the in-mean risk parameter and the threshold/asymmetry parameter.
3. Plot its conditional volatility.
4. Compare its volatility path with the symmetric GARCH(1,1) from Exercise 1.

**Question:** Does allowing leverage materially change the periods identified as high-volatility periods?


```python
from arch.univariate import ARCHInMean, GARCH, StudentsT

# S&P 500 daily returns in percent
y = 100 * spx["Return"].dropna()

# GARCH-in-Mean with GJR-GARCH(1,1) volatility
gjr_m_model = ARCHInMean(
    y,
    lags=5,
    volatility=GARCH(p=1, o=1, q=1),
    distribution=StudentsT(),
    rescale=False
)

# Estimate the model
gjr_m_results = gjr_m_model.fit(
    disp="on",
    options={
        "maxiter": 5000,
        "ftol": 1e-8
    }
)

# Display results
print(gjr_m_results.summary())
```
